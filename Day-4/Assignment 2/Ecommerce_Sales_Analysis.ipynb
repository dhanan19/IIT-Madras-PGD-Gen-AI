```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_excel('ecommerce_orders_raw.xlsx')
df.head()
```

## Part B – Data Cleaning

```python
# Q7 Handle missing ratings
df['Rating'] = df['Rating'].fillna(df['Rating'].median())
df['Rating'].isnull().sum()
```

```python
# Q8 Standardize city names
df['City'] = df['City'].astype(str).str.strip().str.title()
df['City'].unique()[:10]
```

```python
# Q9 Identify invalid quantities
df[df['Quantity'] <= 0]
```

```python
# Q10 Handle duplicate orders
df = df.drop_duplicates()
df.shape
```

```python
# Q11 Convert Order_Date to datetime
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')
df['Order_Date'].dtype
```

## Part C – Feature Engineering

```python
# Q12 Gross_Amount
df['Gross_Amount'] = df['Quantity'] * df['Unit_Price']
df[['Quantity', 'Unit_Price', 'Gross_Amount']].head()
```

```python
# Q13 Discount_Amount
df['Discount_Amount'] = df['Gross_Amount'] * df['Discount'] / 100
df[['Gross_Amount', 'Discount', 'Discount_Amount']].head()
```

```python
# Q14 Net_Amount
df['Net_Amount'] = df['Gross_Amount'] - df['Discount_Amount']
df[['Gross_Amount', 'Discount_Amount', 'Net_Amount']].head()
```

```python
# Q15 Rating_Category
def rating_category(r):
    if r <= 2:
        return 'Poor'
    elif r == 3:
        return 'Average'
    elif r == 4:
        return 'Good'
    else:
        return 'Excellent'

df['Rating_Category'] = df['Rating'].apply(rating_category)
df[['Rating', 'Rating_Category']].head()
```

```python
# Q16 Extract Year, Month, Day, Day of Week
df['Year'] = df['Order_Date'].dt.year
df['Month'] = df['Order_Date'].dt.month
df['Day'] = df['Order_Date'].dt.day
df['Day_of_Week'] = df['Order_Date'].dt.day_name()

df[['Order_Date', 'Year', 'Month', 'Day', 'Day_of_Week']].head()
```

## Part D – Business Analysis
