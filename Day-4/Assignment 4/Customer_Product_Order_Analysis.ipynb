# Assignment 4 – Customer, Product & Order Analysis

```python
import pandas as pd
import numpy as np

# Load datasets
customers = pd.read_csv('customers.csv')
products = pd.read_csv('products.csv')
orders = pd.read_csv('orders.csv')

customers.head(), products.head(), orders.head()
```

## Part A – Understanding Individual Datasets

```python
# Q1 Load all three CSV files
customers, products, orders
```

```python
# Q2 Display dimensions
print('Customers:', customers.shape)
print('Products:', products.shape)
print('Orders:', orders.shape)
```

```python
# Q3 Identify columns
print(customers.columns.tolist())
print(products.columns.tolist())
print(orders.columns.tolist())
```

```python
# Q4 Common columns
common_cp = set(customers.columns).intersection(products.columns)
common_co = set(customers.columns).intersection(orders.columns)
common_po = set(products.columns).intersection(orders.columns)

print('Customers & Products:', common_cp)
print('Customers & Orders:', common_co)
print('Products & Orders:', common_po)
```

```python
# Q5 Relationship between datasets
print('Orders references Customer_ID and Product_ID')
print('Customers is linked to Orders using Customer_ID')
print('Products is linked to Orders using Product_ID')
```

## Part B – Data Cleaning

```python
# Q6 Missing values
print(customers.isnull().sum())
print(products.isnull().sum())
print(orders.isnull().sum())
```

```python
# Q7 Duplicate records
print(customers.duplicated().sum())
print(products.duplicated().sum())
print(orders.duplicated().sum())
```

```python
# Q8 Validity of Customer_ID and Product_ID
invalid_customers = orders[~orders['Customer_ID'].isin(customers['Customer_ID'])]
invalid_products = orders[~orders['Product_ID'].isin(products['Product_ID'])]

print('Invalid Customer IDs:', len(invalid_customers))
print('Invalid Product IDs:', len(invalid_products))
```

```python
# Q9 Convert Order_Date into datetime
orders['Order_Date'] = pd.to_datetime(orders['Order_Date'], errors='coerce')
orders['Order_Date'].dtype
```

```python
# Q10 Every order corresponds to a valid customer
orders['Customer_ID'].isin(customers['Customer_ID']).all()
```

```python
# Q11 Every order corresponds to a valid product
orders['Product_ID'].isin(products['Product_ID']).all()
```

## Part C – Combining Data

```python
# Q12 Merge orders with customers
orders_customers = orders.merge(customers, on='Customer_ID', how='left')
orders_customers.head()
```

```python
# Q13 Merge with products
final_df = orders_customers.merge(products, on='Product_ID', how='left')
final_df.head()
```

```python
# Q14 Final consolidated DataFrame
final_df
```

## Part D – Business Calculations

```python
# Q15 Order_Value
final_df['Order_Value'] = final_df['Price'] * final_df['Quantity']
final_df[['Price', 'Quantity', 'Order_Value']].head()
```

```python
# Q16 Total revenue
final_df['Order_Value'].sum()
```

```python
# Q17 Average order value
final_df['Order_Value'].mean()
```

```python
# Q18 Revenue by product category
final_df.groupby('Category')['Order_Value'].sum().sort_values(ascending=False)
```

```python
# Q19 Revenue by city
final_df.groupby('City')['Order_Value'].sum().sort_values(ascending=False)
```

```python
# Q20 Top 10 customers by spending
final_df.groupby('Customer_Name')['Order_Value'].sum().sort_values(ascending=False).head(10)
```

```python
# Q21 Top 10 products by revenue
final_df.groupby('Product_Name')['Order_Value'].sum().sort_values(ascending=False).head(10)
```

```python
# Q22 Most frequently purchased product
final_df.groupby('Product_Name')['Quantity'].sum().sort_values(ascending=False).head(1)
```

```python
# Q23 City with the highest number of customers
customers['City'].value_counts().head(1)
```

```python
# Q24 City generating the highest revenue
city_revenue = final_df.groupby('City')['Order_Value'].sum()
city_revenue.idxmax(), city_revenue.max()
```

## Part E – Customer Analysis

```python
# Q25 Age groups
final_df['Age_Group'] = pd.cut(
    final_df['Age'],
    bins=[17, 30, 45, 60, np.inf],
    labels=['18-30', '31-45', '46-60', '61+']
)
final_df[['Age', 'Age_Group']].head()
```

```python
# Q26 Revenue by age group
final_df.groupby('Age_Group', observed=False)['Order_Value'].sum()
```

```python
# Q27 Spending by gender
final_df.groupby('Gender')['Order_Value'].sum()
```

```python
# Q28 High-value customers (top 10% by spending)
customer_spending = final_df.groupby('Customer_Name')['Order_Value'].sum()
threshold = customer_spending.quantile(0.90)
customer_spending[customer_spending >= threshold].sort_values(ascending=False)
```

```python
# Q29 Percentage of customers with more than one order
orders_per_customer = final_df.groupby('Customer_ID')['Order_ID'].nunique()
percentage = (orders_per_customer.gt(1).mean()) * 100
percentage
```

## Part F – Advanced Analysis

```python
# Q30 Customer spending ranking
customer_ranking = (
    customer_spending
    .sort_values(ascending=False)
    .rank(method='dense', ascending=False)
)
customer_ranking
```

```python
# Q31 Top 10% of customers based on spending
top_10_percent = customer_spending[customer_spending >= threshold].sort_values(ascending=False)
top_10_percent
```

```python
# Q32 Most profitable product category for each city
city_category = (
    final_df.groupby(['City', 'Category'])['Order_Value']
    .sum()
    .reset_index()
)
city_category.loc[city_category.groupby('City')['Order_Value'].idxmax()]
```

```python
# Q33 Most popular product in each category
category_product = (
    final_df.groupby(['Category', 'Product_Name'])['Quantity']
    .sum()
    .reset_index()
)
category_product.loc[category_product.groupby('Category')['Quantity'].idxmax()]
```

```python
# Q34 Monthly revenue
final_df['Month'] = final_df['Order_Date'].dt.to_period('M')
monthly_revenue = final_df.groupby('Month')['Order_Value'].sum()
monthly_revenue
```

```python
# Q35 Month with the highest revenue
monthly_revenue.idxmax(), monthly_revenue.max()
```
