```python
import pandas as pd
import numpy as np

df = pd.read_excel('patient_clinical_data_raw.xlsx')
df.head()
```

```python
# Q5 Statistical summary of numerical columns
df.describe()
```

## Part B – Data Cleaning

```python
# Q7 Columns containing missing values
df.isnull().sum()[df.isnull().sum() > 0]
```

```python
# Q8 Percentage of missing values
(df.isnull().mean() * 100).round(2)
```

```python
# Q9 Handle missing values in numerical columns
num_cols = df.select_dtypes(include='number').columns
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
df[num_cols].isnull().sum()
```

```python
# Q10 Remove duplicate records
df = df.drop_duplicates()
df.shape
```

```python
# Q11 Check the Age column for invalid values
df[(df['Age'] < 0) | (df['Age'] > 120)]
```

```python
# Q12 Handle invalid ages by replacing with median age
median_age = df.loc[df['Age'].between(0, 120), 'Age'].median()
df.loc[(df['Age'] < 0) | (df['Age'] > 120), 'Age'] = median_age
```

```python
# Q13 Identify inconsistent categorical values
for col in ['Gender', 'Department']:
    print(col)
    print(df[col].unique())
    print()
```

```python
# Q14 Standardize categorical values
df['Gender'] = (
    df['Gender']
    .astype(str)
    .str.strip()
    .str.lower()
    .replace({'m': 'Male', 'male': 'Male', 'f': 'Female', 'female': 'Female'})
)

df['Department'] = df['Department'].astype(str).str.strip().str.title()

df[['Gender', 'Department']].head()
```

## Part C – Data Transformation

```python
# Q15 Create Age_Group
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[-1, 17, 40, 60, np.inf],
    labels=['Child', 'Young Adult', 'Middle Age', 'Senior']
)
df[['Age', 'Age_Group']].head()
```

```python
# Q16 Separate Blood_Pressure into Systolic and Diastolic
bp = df['Blood_Pressure'].astype(str).str.split('/', expand=True)
df['Systolic'] = pd.to_numeric(bp[0], errors='coerce')
df['Diastolic'] = pd.to_numeric(bp[1], errors='coerce')
df[['Blood_Pressure', 'Systolic', 'Diastolic']].head()
```

```python
# Q17 Create High_Risk column
df['High_Risk'] = (
    (df['Age'] > 60) |
    (df['Systolic'] >= 140) |
    (df['Diastolic'] >= 90) |
    (df['Glucose'] >= 140) |
    (df['Cholesterol'] >= 240)
)
df[['Age', 'Systolic', 'Diastolic', 'Glucose', 'Cholesterol', 'High_Risk']].head()
```

```python
# Q18 Create Cost_Per_Day
df['Cost_Per_Day'] = df['Treatment_Cost'] / df['Length_of_Stay'].replace(0, np.nan)
df[['Treatment_Cost', 'Length_of_Stay', 'Cost_Per_Day']].head()
```

## Part D – Analysis

```python
# Q19 Average age
df['Age'].mean()
```

```python
# Q20 Number of patients in each department
df['Department'].value_counts()
```

```python
# Q21 Average treatment cost for each department
df.groupby('Department')['Treatment_Cost'].mean()
```

```python
# Q22 Average length of stay for each diagnosis
df.groupby('Diagnosis')['Length_of_Stay'].mean()
```

```python
# Q23 Top 10 patients with highest treatment cost
df.nlargest(10, 'Treatment_Cost')[['Patient_ID', 'Treatment_Cost']]
```

```python
# Q24 Department with highest average treatment cost
dept_avg = df.groupby('Department')['Treatment_Cost'].mean()
dept_avg.idxmax(), dept_avg.max()
```

```python
# Q25 Percentage of patients who were readmitted
(df['Readmission'].astype(str).str.lower().eq('yes').mean() * 100)
```

```python
# Q26 Readmission rates across departments
(
    df.assign(Readmission=df['Readmission'].astype(str).str.lower().eq('yes'))
      .groupby('Department')['Readmission']
      .mean() * 100
)
```

```python
# Q27 Average cholesterol and glucose level by diagnosis
df.groupby('Diagnosis')[['Cholesterol', 'Glucose']].mean()
```

```python
# Q28 High glucose patients
high_glucose = df[df['Glucose'] >= 140]

# High cholesterol patients
high_cholesterol = df[df['Cholesterol'] >= 240]

# High heart rate patients
high_heart_rate = df[df['Heart_Rate'] >= 100]

print('High glucose:')
print(high_glucose[['Patient_ID', 'Glucose']].head())

print('\nHigh cholesterol:')
print(high_cholesterol[['Patient_ID', 'Cholesterol']].head())

print('\nHigh heart rate:')
print(high_heart_rate[['Patient_ID', 'Heart_Rate']].head())
```
