```python
ratings = {
    "Movie A": 4.5,
    "Movie B": None,
    "Movie C": 3.8,
    "Movie D": None
}
```

```python
# Questions

Check whether Movie C has a rating.
Find all movies with no ratings.
Count unrated movies.
Replace all missing ratings with 0.
```

```python
# Check whether Movie C has a rating 
if ratings["Movie C"] is not None: 
    print("Movie C has a rating")
```

```python
#Q17. Find the employee with the highest salary
highest = max(employees, key=lambda x: employees[x]["Salary"])

print(highest, employees[highest])
```

```python
#Q18. Find the employee with the lowest salary
lowest = min(employees, key=lambda x: employees[x]["Salary"])

print(lowest, employees[lowest])
```

```python
#Q19. Calculate the average salary of all employees
total = sum(details["Salary"] for details in employees.values())

average = total / len(employees)

print(average)
```

```python
#Q22. Count how many employees belong to each department
dept_count = {}

for details in employees.values():
    dept = details["Department"]
    dept_count[dept] = dept_count.get(dept, 0) + 1

for dept, count in dept_count.items():
    print(dept, ":", count)
```

```python
#Q23. Count how many employees are managed by each manager
manager_count = {}

for details in employees.values():
    manager = details["Manager"]
    manager_count[manager] = manager_count.get(manager, 0) + 1

for manager, count in manager_count.items():
    print(manager, ":", count)
```

```python
#Q24. Find which department has the largest number of employees
dept_count = {}

for details in employees.values():
    dept = details["Department"]
    dept_count[dept] = dept_count.get(dept, 0) + 1

largest = max(dept_count, key=dept_count.get)

print(largest, dept_count[largest])
```
