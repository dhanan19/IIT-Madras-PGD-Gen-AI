## Task 1 — Create a Tensor
Create a 1D tensor containing 25.2, 25.5, 24.9, 26.1, 25.8.

```python
import numpy as np
sensor = np.array([25.2, 25.5, 24.9, 26.1, 25.8])
print("Tensor:", sensor)
print("Shape:", sensor.shape)
print("Dimensions:", sensor.ndim)
print("Elements:", sensor.size)
```

Expected: shape `(5,)`, 1 dimension, 5 elements.

## Task 2 — From Vector to Matrix
Rows = time, columns = sensors.

```python
temperature = [25, 26, 25, 27, 28]
pressure = [101, 102, 101, 103, 104]
humidity = [60, 62, 61, 63, 64]
vibration = [0.2, 0.3, 0.2, 0.4, 0.5]
X = np.array([temperature, pressure, humidity, vibration]).T
print(X)
print("Shape:", X.shape)
print("Dimensions:", X.ndim)
print("First time step:", X[0])
print("Temperature:", X[:, 0])
print("Vibration:", X[:, 3])
```

Expected shape `(5, 4)`. Axis 0 is time; axis 1 is sensor.

## Task 3 — Create a 3D Tensor
Interpretation: Machine × Time × Sensor.

```python
X3 = np.arange(10 * 5 * 4).reshape(10, 5, 4)
print("Shape:", X3.shape)

# Machine 3 = index 2, time step 2 = index 1, temperature = sensor index 0
print("Machine 3, time 2, temperature:", X3[2, 1, 0])
print("Machine 5:", X3[4])
print("All machines at time 1:", X3[:, 0, :])
print("Sensor 4 from all machines/times:", X3[:, :, 3])
```

Shapes: machine 5 `(5,4)`, all machines at time 1 `(10,4)`, sensor 4 `(10,5)`. NumPy uses zero-based indexing.

## Task 4 — Indexing a Tensor

```python
X = np.arange(24).reshape(2, 3, 4)
print(X)
print("X.shape:", X.shape)
print("X[0]:", X[0])
print("X[1]:", X[1])
print("X[0,0]:", X[0,0])
print("X[0,0,0]:", X[0,0,0])
```

`X[0]` is the first block `(3,4)`. `X[0,0]` is its first row `(4,)`. `X[0,0,0]` is one scalar.

## Task 8 — Reshape

```python
X = np.arange(24)
for shape in [(4,6), (2,3,4), (6,4), (24,)]:
    Y = X.reshape(shape)
    print("\nShape:", Y.shape)
    print("Dimensions:", Y.ndim)
    print(Y)
```

Every reshape contains 24 elements. Reshape changes the organization/shape, not the total number of elements.

## Task 9 — Reshape vs Transpose

```python
X = np.arange(24).reshape(2,3,4)
reshaped = X.reshape(3,2,4)
transposed = X.transpose(1,0,2)
print("Reshape:\n", reshaped)
print("Transpose:\n", transposed)
print("Same values in same positions?", np.array_equal(reshaped, transposed))
```

They are not conceptually the same: `reshape` reorganizes elements into a new shape; `transpose` changes axis order.

## Task 10 — Broadcasting

```python
X = np.ones((3,4))
b = np.array([10,20,30,40])
print("X + b:\n", X + b)
print("X * b:\n", X * b)
```

A `(4,)` vector broadcasts across the last dimension of `(3,4)`, so the same four values are applied to every row.

## Task 11 — Broadcasting with Sensor Data

```python
X = np.random.randn(100,10)
calibration = np.array([1.2,0.8,1.1,0.5,0.9,1.0,0.7,1.3,0.6,0.4])
corrected = X - calibration
print("X:", X.shape)
print("Calibration:", calibration.shape)
print("Corrected:", corrected.shape)
```

The output remains `(100,10)` because one calibration value is broadcast to each corresponding sensor column.

## Task 12 — Tensor Aggregation

```python
X = np.random.randn(100,20,5)
mean_time = X.mean(axis=1)
mean_sensors = X.mean(axis=2)
mean_machines = X.mean(axis=0)
max_time = X.max(axis=1)
std_time = X.std(axis=1)
print("Mean across time:", mean_time.shape)
print("Mean across sensors:", mean_sensors.shape)
print("Mean across machines:", mean_machines.shape)
print("Max across time:", max_time.shape)
print("Std across time:", std_time.shape)
```

Expected shapes: mean-time `(100,5)`, mean-sensors `(100,20)`, mean-machines `(20,5)`, max-time `(100,5)`, std-time `(100,5)`.

## Task 13 — Normalize a Tensor

```python
X = np.random.randn(100,20,5)
mean = X.mean(axis=(0,1), keepdims=True)
std = X.std(axis=(0,1), keepdims=True)
X_normalized = (X - mean) / std
print("Normalized shape:", X_normalized.shape)
print("Mean per sensor:", X_normalized.mean(axis=(0,1)))
print("Std per sensor:", X_normalized.std(axis=(0,1)))
```

After normalization, each sensor should have mean approximately 0 and standard deviation approximately 1.

## Task 14 — Flattening

```python
image = np.random.rand(28,28,3)
flat_image = image.reshape(2352)
batch = np.random.rand(64,28,28,3)
flat_batch = batch.reshape(64,2352)
print("Image:", image.shape, "->", flat_image.shape)
print("Batch:", batch.shape, "->", flat_batch.shape)
```

A single image has `28×28×3 = 2352` elements. For a batch, keep the batch axis: `(64,2352)`. Flattening everything to `(150528,)` would lose the separation between the 64 images.

## Task 15 — Concatenation

```python
visual_features = np.random.randn(32,128)
sensor_features = np.random.randn(32,64)
combined = np.concatenate([visual_features, sensor_features], axis=1)
print("Visual:", visual_features.shape)
print("Sensor:", sensor_features.shape)
print("Combined:", combined.shape)
```

Result: `(32,192)` because `128+64=192`. Concatenation along the feature axis is possible because both tensors have 32 samples.

## Task 16 — Practical Mini Challenge

```python
X = np.random.randn(500,60,8)

machine_10 = X[9]
first_20 = X[:, :20, :]
sensor_5 = X[:, :, 4]
average_sensor = X.mean(axis=(0,1))
average_time = X.mean(axis=(0,2))

machine_average = X.mean(axis=(1,2))
best_machine = np.argmax(machine_average) + 1

sensor_mean = X.mean(axis=(0,1), keepdims=True)
sensor_std = X.std(axis=(0,1), keepdims=True)
X_normalized = (X - sensor_mean) / sensor_std

X_train = X[:400]
X_test = X[400:]

print("Dataset:", X.shape)
print("Machine 10:", machine_10.shape)
print("First 20 time steps:", first_20.shape)
print("Sensor 5:", sensor_5.shape)
print("Average each sensor:", average_sensor.shape)
print("Average each time step:", average_time.shape)
print("Highest-average machine:", best_machine)
print("Normalized:", X_normalized.shape)
print("Training:", X_train.shape)
print("Testing:", X_test.shape)
```

Expected shapes: dataset `(500,60,8)`, machine 10 `(60,8)`, first 20 `(500,20,8)`, sensor 5 `(500,60)`, sensor averages `(8,)`, time averages `(60,)`, training `(400,60,8)`, testing `(100,60,8)`. The highest-average machine depends on random data.
