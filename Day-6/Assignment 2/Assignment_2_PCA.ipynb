## Question 10

```python
X = np.array([[2,1],[3,2],[4,3],[5,4],[6,5]])
mean = X.mean(axis=0)
X_centered = X - mean
cov = np.cov(X_centered, rowvar=False)
vals, vecs = np.linalg.eigh(cov)
order = np.argsort(vals)[::-1]
vals, vecs = vals[order], vecs[:,order]
components = vecs[:, :2]
transformed = X_centered @ components
print("Mean:", mean)
print("Centered data:\n", X_centered)
print("Covariance:\n", cov)
print("Sorted eigenvalues:", vals)
print("Principal directions:\n", vecs)
print("Transformed data:\n", transformed)
```

## Question 11

```python
X = np.array([[2,1],[3,2],[4,3],[5,4],[6,5]])
Xc = X - X.mean(axis=0)
cov = np.cov(Xc, rowvar=False)
vals, vecs = np.linalg.eigh(cov)
pc1 = vecs[:, np.argmax(vals)]
projected = Xc @ pc1
print("PC1:", pc1)
print("1D representation:\n", projected.reshape(-1,1))
```

## Question 12

```python
def pca_from_scratch(X, n_components):
    Xc = X - X.mean(axis=0)
    cov = np.cov(Xc, rowvar=False)
    vals, vecs = np.linalg.eigh(cov)
    order = np.argsort(vals)[::-1]
    vals, vecs = vals[order], vecs[:,order]
    components = vecs[:, :n_components]
    return Xc @ components, vals, components

X = np.array([[2,1],[3,2],[4,3],[5,4],[6,5]])
n_components = 1
reduced, vals, components = pca_from_scratch(X, n_components)
print("Selected eigenvector:\n", components)
print("Reduced data:\n", reduced)
```

```python
from sklearn.preprocessing import StandardScaler
X = np.array([[1,100],[2,200],[3,300],[4,400],[5,500]])

pca1 = PCA(n_components=2)
without_scaling = pca1.fit_transform(X)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
pca2 = PCA(n_components=2)
with_scaling = pca2.fit_transform(X_scaled)

print("Without standardization:\n", without_scaling)
print("Explained variance ratio without standardization:", pca1.explained_variance_ratio_)
print("\nStandardized data:\n", X_scaled)
print("PCA after standardization:\n", with_scaling)
print("Explained variance ratio after standardization:", pca2.explained_variance_ratio_)
print("\nObservation: without scaling, the larger-scale feature dominates variance; standardization puts features on comparable scales.")
```
